```python
fruits=['apple','banana','cherry']
```

```python
fruits.extend([
   'orange','grape','mange','pineapple''strawberry',
   'blueberry','watermelon','peach','kiwi','pear'
])
```

```python
fruits
```

```python
for a in fruits:
  print(a)
```

```python
upper_fruits=[]
for fruit in fruits:
  upper_fruits.append(fruit.upper())
print(upper_fruits)
```

```python
long_fruits=[]
for fruit in fruits:
  if len(fruit)>=6:
    long_fruits.append(fruit)
print(long_fruits)
```

```python
for i,fruits in enumerate(fruits):
  print(f"{i+1}.{fruit}")
```

```python
joseon_kings = [
    "태조", "정종", "태종", "세종", "문종", "단종", "세조", "예종", "성종",
    "연산군", "중종", "인종", "명종", "선조", "광해군", "인조", "효종", "현종",
    "숙종", "경종", "영조", "정조", "순조", "헌종", "철종", "고종", "순종"
]
```

```python
for i,a in enumerate(joseon_kings):
  print(f'{i+1}대:{a}')
```

```python
n=int(input('몇 년대 왕 알고 싶어'))
if 1 <=n<=len(joseon_kings):
  king_name=joseon_kings[n-1]
  print(f"{n}대 왕은 {king_name}입니다")
else:
  print(f"잘못됭 입력입니다")
```

```python
name=input("왕 이름을 입력하세요:")
if name in joseon_kings:
  index=joseon_kings.index(name)
  print(f"{name}은 {index+1}번째 왕입니다")
else:
  print("해당 왕이 업습니다")
```

```python
rev=reversed(joseon_kings)
for r in rev:
  print(r,end="")
```
