```python
import random
answer=random.randint(1,100)
```

```python
answer
```

```python
user=int(input("추측값을 입력하시오"))
```

```python
count=0
while user!=answer:
  if user<answer:
    print("더 큰 수를 입력하시오")
    count+=1
  else:
    print("더 작은 수를 입력하시오")
  user=int(input("추측값을 입력하시오"))
print("총 시도 횟수:",count)
```

```python
num=int(input("숫자를 입력하세요"))
print(f'{num}의 약수는 다음과 같습니다:')
for i in range(1,num+1):
  if num%i==0:
    print(i,end="")
```
