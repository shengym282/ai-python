```python
import random
answer=random.randint(1,100)
user=int(input("추측값을 입력하시오:"))
count=1
while user != answer:
  if user < answer:
    print("더 큰 수를 입력하시오")
  else:
    print("더 작은 수를 입력하시오")
  user=int(input("추측값을 입력하시오:"))
  count +=1

print(f"정답입니다!총 시도 횟수:{count}")
```
