```python
cafe_menu= {
    "아메리카노": 4500,
    "카페라떼": 5200,
    "바닐라라떼": 5800,
    "카푸치노": 5500,
    "카라멜 마키아토": 5900,
    "콜드브루": 5500,
    "카페모카": 6000,
    "말차라떼": 6200,
    "레몬에이드": 6500,
    "딸기스무디": 6900}
sum=0

while True:
  a=input('음료 종류 또는 종료입럭')
  if a=='종료':
    break
  elif a in cafe_menu:
    n=int(input(f'{a}주문 수：'))
    sum +=cafe_menu[a]*n
  else:
    print(f'{a}는 업습니다.')
print(f'총 금액:{sum}원')
```

```python
for key,value in cafe_menu.items():
  print(f'{key}의 가격은 {value}원입니다')
```
