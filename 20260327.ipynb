```python
import random

dice1=random.randint(1,6)
dice2=random.randint(1,6)
print(f'주사위1:{dice1}')
print(f'주사위2:{dice2}')
if dice1>dice2:
  print("dice1 더 크다")
elif dice1==dice2:
  print("dice1과 dice2 같다")
else:
  print("dice2 더 크다")
```
